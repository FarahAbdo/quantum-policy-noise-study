```python
"""
Live visualization — شاهدي الـ policy وهي بتلعب CartPole في الوقت الفعلي.

بنعرض:
- العربية والعصاية بتتحرك (نافذة الـ Gym)
- الـ action probabilities في كل خطوة
- الـ reward بيتراكم
- شريط بياني للـ rewards عبر الـ episodes

كرّري ده مرة قبل التدريب ومرة بعده وشوفي الفرق.

USAGE:
    python watch_live.py           # quantum policy عشوائي (قبل التدريب)
    python watch_live.py trained   # هيدور على params مدربة
"""

import sys
import os
import time
import numpy as np
import gymnasium as gym
import matplotlib
matplotlib.use("TkAgg")  # عشان النوافذ تشتغل على ويندوز
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.dirname(os.path.abspath(__file__)))
from src.quantum_policy import make_policy, action_probs, init_params


# ---- الإعدادات ----
N_EPISODES = 5         # كم episode نتفرّج عليه
RENDER_DELAY = 0.02    # ثواني بين الـ frames (زوّدها لو عايزة تشوفي ببطء)


def run_one_episode(env, circuit, params, episode_idx, ax_probs, ax_rewards, reward_history):
    """
    شغّلي episode واحد مع رسم مباشر للـ action probabilities والـ reward.
    """
    state, _ = env.reset(seed=episode_idx)
    total_reward = 0.0
    step = 0
    step_rewards = []
    step_probs = []

    print(f"\n{'='*55}")
    print(f"Episode {episode_idx + 1}")
    print(f"{'='*55}")
    print(f"{'Step':>4} {'P(right)':>10} {'P(left)':>10} {'Action':>8} {'Reward':>8}")
    print("-"*55)

    done = False
    while not done and step < 500:
        # 1. الـ policy تشوف الـ state وتطلع احتمالات
        probs = action_probs(circuit, state, params)

        # 2. اختاري action بشكل عشوائي حسب الاحتمالات
        action = int(np.random.choice(2, p=probs))

        # 3. الـ environment ينفذ الـ action ويرد بالحالة الجديدة
        state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated

        total_reward += reward
        step += 1
        step_rewards.append(total_reward)
        step_probs.append(probs)

        # طبعي كل 5 خطوات
        if step % 5 == 0 or done:
            action_name = "يمين →" if action == 0 else "← شمال"
            print(f"{step:>4} {probs[0]:>10.3f} {probs[1]:>10.3f} {action_name:>10} {total_reward:>8.0f}")

        # تأخير صغير عشان نقدر نشوف الحركة
        time.sleep(RENDER_DELAY)

    print("-"*55)
    print(f"  ✅ Episode {episode_idx + 1} انتهى بـ reward = {total_reward:.0f}")

    # تحديث الرسومات
    reward_history.append(total_reward)

    # رسمة 1: الاحتمالات خلال الـ episode الأخير
    ax_probs.clear()
    step_probs = np.array(step_probs)
    ax_probs.plot(step_probs[:, 0], label="P(right)", color="tab:blue")
    ax_probs.plot(step_probs[:, 1], label="P(left)", color="tab:red")
    ax_probs.axhline(0.5, color="gray", linestyle="--", alpha=0.5)
    ax_probs.set_xlabel("Step")
    ax_probs.set_ylabel("Probability")
    ax_probs.set_title(f"Action probs خلال Episode {episode_idx + 1}")
    ax_probs.legend()
    ax_probs.set_ylim(0, 1)
    ax_probs.grid(alpha=0.3)

    # رسمة 2: الـ rewards كل الـ episodes
    ax_rewards.clear()
    ax_rewards.bar(range(1, len(reward_history) + 1), reward_history,
                   color="tab:green", edgecolor="black")
    ax_rewards.set_xlabel("Episode")
    ax_rewards.set_ylabel("Total Reward")
    ax_rewards.set_title("Reward لكل episode")
    ax_rewards.grid(alpha=0.3, axis="y")
    for i, r in enumerate(reward_history):
        ax_rewards.text(i + 1, r + 1, f"{r:.0f}", ha="center", fontsize=9)

    plt.pause(0.5)
    return total_reward


def main():
    use_trained = len(sys.argv) > 1 and sys.argv[1] == "trained"

    # ابني الـ circuit (بدون noise عشان السرعة)
    circuit = make_policy(noise_prob=0.0)

    # حملي الـ params
    if use_trained and os.path.exists("results/trained_params.npy"):
        params = np.load("results/trained_params.npy")
        print("✅ تم تحميل params مدربة من results/trained_params.npy")
    else:
        params = init_params(seed=42)
        print("⚠️  بنستخدم params عشوائية (مش مدربة)")
        print("   ده يعني الـ policy هتلعب random تقريباً")

    # ابني الـ environment مع render mode
    env = gym.make("CartPole-v1", render_mode="human")

    # ابني نافذة الرسومات
    plt.ion()
    fig, (ax_probs, ax_rewards) = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle("Quantum Policy — Live Visualization",
                 fontsize=14, fontweight="bold")

    reward_history = []

    try:
        for ep in range(N_EPISODES):
            run_one_episode(env, circuit, params, ep,
                            ax_probs, ax_rewards, reward_history)

        print("\n" + "="*55)
        print("✨ الخلاصة")
        print("="*55)
        print(f"  متوسط الـ reward: {np.mean(reward_history):.1f}")
        print(f"  أحسن episode:    {np.max(reward_history):.0f}")
        print(f"  أسوأ episode:    {np.min(reward_history):.0f}")
        print("\nاضغطي على الرسمة عشان تقفليها.")
        plt.ioff()
        plt.show()
    finally:
        env.close()


if __name__ == "__main__":
    main()
```
